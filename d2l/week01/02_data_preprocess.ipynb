```python
import pandas as pd
import numpy as np
import torch
def load_csv(path):
    date =pd.read_csv(path)
    inputs,outputs = date.iloc[ : ,0:-1],date.iloc[ : ,-1]
    inputs =pd.get_dummies( inputs.fillna(inputs.mean(numeric_only=True)) , dummy_na=True)
    x = torch.tensor(inputs.to_numpy(dtype=np.float32))
    y = torch.tensor(outputs.to_numpy(dtype=np.float32))
    return x,y
x, y = load_csv('../data/house_tiny.csv')
print(x.shape, x.dtype)
print(y.shape, y.dtype)
print(x)
```
